```python
import os 
import fastf1
import pandas as pd
import json
```

```python

os.makedirs("cache", exist_ok=True)
fastf1.Cache.enable_cache("cache")

session = fastf1.get_session(2024, "Bahrain", "R")

session.load()
```

```python
# Filter and print all available attributes that hold data or DataFrames
attributes = [attr for attr in dir(session) if not attr.startswith('_')]

for attr in attributes:
    # Check if the attribute is a pandas DataFrame or similar data container
    obj = getattr(session, attr)
    if hasattr(obj, 'shape') or isinstance(obj, (dict, list)):
        print(f"session.{attr}")
```

```python
# binding tables datas for its names in python in pandas
car_data = session.car_data
driver = session.drivers
event = session.event
f1_api = session.f1_api_support
lap = session.laps
position = session.pos_data
control_sms = session.race_control_messages
result = session.results
session_info = session.session_info
session_status = session.session_status
track_status = session.track_status
weather = session.weather_data
```

```python
# Created the  dictionary or list of my bound variables with their names
variables = {
    "car_data": car_data,
    "driver": driver,
    "event": event,
    "f1_api": f1_api,
    "lap": lap,
    "position": position,
    "control_sms": control_sms,
    "result": result,
    "session_info": session_info,
    "session_status": session_status,
    "track_status": track_status,
    "weather": weather
}

# Print the name and the exact data type for each one
for name, var in variables.items():
    print(f"{name:15} -> {type(var)}")
```
